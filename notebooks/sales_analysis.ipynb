## Business Context

### 1.1 Business Problem

Our e-commerce operation is growing and processing an increasing volume os daily transactions. However, the sales data is still being used mostly in its raw form, limiting the ability to extract meaningful insights and suppport strategic decisions.

This lack of visibility creates several business challenges:

- **Inefficient Inventory Management:** There is limited clarity about which products have the highest demand and which products have low turnover. This can lead to excess inventory, stockouts, and unnecessary storage costs.

- **Low Marketing Efficiency:** Marketing actions are not sufficiently supported by customer purchasing patterns, product category performance, or geographic concentration of sales.

- **Unclear Expansio Priorities:** The business does not have a clear view of which regions represent the strongest commercial opportunities or where future expansio efforts should be prioritized.

The main business problem is the lack of a structured analytical view of sales performance, limiting the company's ability to make faster and more data-driven decisions.

### 1.2 Project Objectives  

The objective of this project is to transform raw sales data into actionable business insights.  

The analisys aims to answer four main business questions:  

- **What should we sell?** Identify the best-performing products and support inventory and portfolio decisions.  

- **Where should we focus?**  Understand which product categories generate the highest revenue.  

- **When should we act?** Analyze sales performance over time to identify trends, peaks, and possible seasonal patterns.

- **Where should we expand?** Evaluate the geographic distribution of sales and identify the strongest markets.  

### 1.3 Proposed Solution  

The proposed solution is to generate, structure, clean and analyze a synthetic e-commerce sales dataset using Python.  

The analytical worksflow will use tools such as Pandas, NumPy, Matplotlib and Seaborn to:

- analyze product, category, temporal and geographic performance;  
- create visualizations;

The final output will be an analytical report designed to support management, marketing and operational decision-making.  

### 1.4 Expected Business Benefits  

The projetc is expected to support the following business outcomes:  

- **Inventory Optimization:** Identify high and low performing products to improve purchasing and stock allocation decisions.  

- **More Targeted Marketing:** Use category and regional performance to support more focused campaigns and improve marketing efficiency.

## 5. Data Cleaning, Preprocessing and Feature Engineering  

In this section, the dataset is cleaned and prepared for analysis. The process includes handling data types, checking for inconsistencies and creating new features that will support the business analysis.

```python
# Feature Engineering: Create a new column 'Total_Sale' which is the product of 'Price' and 'Quantity_Sold'
sales_df['Total_Sale'] = sales_df['Price'] * sales_df['Quantity_Sold']

# Using lambda function to create a new column 'Status_Received' based on the 'States' column
sales_df['Status_Received'] = sales_df['State'].apply(lambda State: 'Flash' if State in ['SP', 'RJ', 'MG'] else 'Normal')

sales_df.head(5)
```

## 6. Analysis 1 - Top 10 Best-Selling Products  
Wich products generated the highest sales volume?

```python
top10_sales_products = sales_df.groupby('Product')['Quantity_Sold'].sum().sort_values(ascending=False).head(10)
top10_sales_products
```

```python
# Define a style for the graphs
sns.set_style("whitegrid")

# Create a figure and axis for the bar plot
plt.figure(figsize=(12, 7))

# Create the horizontal bar plot
ax = top10_sales_products.sort_values(ascending=True).plot(
    kind="barh",
    color="skyblue"
)

# Add data labels
ax.bar_label(ax.containers[0], padding=3, fmt="%.0f")

# Add title and labels to the graph
plt.title("Top 10 Products by Quantity Sold", fontsize=16)
plt.xlabel("Quantity Sold", fontsize=14)
plt.ylabel("Product", fontsize=14)

# Adjust layout and display the graph
plt.tight_layout()
plt.show()
```
