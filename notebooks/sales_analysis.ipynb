## 5. Data Cleaning, Preprocessing and Feature Engineering  

In this section, the dataset is cleaned and prepared for analysis. The process includes handling data types, checking for inconsistencies and creating new features that will support the business analysis.

```python
# Ensure Sale_Date is stored as datetime
sales_df['Sale_Date'] = pd.to_datetime(sales_df['Sale_Date'])
```

```python
# Feature engineering: calculate total revenue per transaction
sales_df['Total_Sale'] = (
    sales_df['Price'] * sales_df['Quantity_Sold']
)
```

```python
# Feature engineering:
# create a delivery status based on the customer's state
sales_df['Delivery_Status'] = sales_df['State'].apply(
    lambda state: 'Fast' if state in ['SP', 'RJ', 'MG'] else 'Standard'
)
```

## 6. Analysis 1 - Top 10 Best-Selling Products  
Wich products generated the highest sales volume?

```python
# Group by product and calculate total quantity sold
top10_sales_products = (
    sales_df
    .groupby('Product')['Quantity_Sold']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top10_sales_products
```

```python
# Define chart style
sns.set_style("whitegrid")

# Create a new figure
plt.figure(figsize=(12, 7))

# Create horizontal bar plot
ax = (
    top10_sales_products
    .sort_values(ascending=True)
    .plot(
        kind='barh',
        color='skyblue'
    )
)

# Add data labels
ax.bar_label(
    ax.containers[0],
    padding=3,
    fmt='%.0f'
)

# Add title and labels
plt.title(
    'Top 10 Products by Quantity Sold',
    fontsize=16
)

plt.xlabel(
    'Quantity Sold',
    fontsize=14
)

plt.ylabel(
    'Product',
    fontsize=14
)

plt.tight_layout()

plt.savefig(
    r"C:\Users\Guilherme\projects_python\sales_data_analysis_ecomerce\images\top10_products_quantity_sold.png",
    dpi=300,
    bbox_inches='tight'
)
plt.show()
```

## 7. Analysis 2 - Monthly Revenue  
How does monthly revenue evolve over time, and are there any visible trends or seasonal patterns?

```python
# Create a month column for monthly aggregation
sales_df['Sale_Month'] = (
    sales_df['Sale_Date']
    .dt
    .to_period('M')
)
```

```python
# Calculate monthly revenue
monthly_sales = (
    sales_df
    .groupby('Sale_Month')['Total_Sale']
    .sum()
)
```

```python
# Convert index to string for visualization
monthly_sales.index = monthly_sales.index.strftime('%Y-%m')
```

```python
# Display formatted monthly revenue
monthly_sales.map('R$ {:,.2f}'.format)
```

```python
# Create a new figure
plt.figure(figsize=(12, 7))

# Plot monthly revenue
ax = monthly_sales.plot(
    kind='line',
    marker='o',
    linestyle='-',
    color='skyblue'
)

# Format Y-axis in millions
ax.yaxis.set_major_formatter(
    FuncFormatter(
        lambda x, pos: f'R$ {x / 1_000_000:.1f}M'
    )
)

# Add data labels
for x, y in enumerate(monthly_sales.values):
    ax.annotate(
        f'R$ {y / 1_000_000:.1f}M',
        (x, y),
        textcoords='offset points',
        xytext=(0, 8),
        ha='center'
    )

# Add title and labels
plt.title(
    'Monthly Revenue Over Time',
    fontsize=16
)

plt.xlabel(
    'Month',
    fontsize=14
)

plt.ylabel(
    'Revenue (BRL)',
    fontsize=14
)

plt.xticks(rotation=45)

plt.grid(
    True,
    which='both',
    linestyle='--',
    linewidth=0.5
)

plt.tight_layout()

plt.savefig(
    r"C:\Users\Guilherme\projects_python\sales_data_analysis_ecomerce\images\monthly_revenue.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

## 8. Analysis 3 - Revenue by State
